# Functions

## Type Casting Functions

### str() Function

```python
x = 5  # Integer data type stored inside variable x
print(str(x))       # int → str
```

```python
x = 3.14  # Float data type stored inside variable x
print(str(x))       # float → str
```

```python
x = True  # Boolean data type stored inside variable x
print(str(x))       # bool → str ('True' or 'False')
```

```python
x = False  # Boolean data type stored inside variable x
print(str(x))       # bool → str ('True' or 'False')
```

```python
x = [1, 2, 3]  # List data type stored inside variable x
print(str(x))       # list → str
```

### int() Function

```python
x = 3.14  # Float data type stored inside variable x
print(int(x))       # float → int (decimal part truncated)
```

```python
x = "123"  # String data type stored inside variable x
print(int(x))       # str → int (only works if string is a valid whole number)
```

```python
x = True  # Boolean data type stored inside variable x
print(int(x))       # bool → int (True → 1, False → 0)
```

```python
x = False  # Boolean data type stored inside variable x
print(int(x))       # bool → int (True → 1, False → 0)
```

### float() Function

```python
x = 5  # Integer data type stored inside variable x
print(float(x))     # int → float
```

```python
x = "123"  # String data type stored inside variable x
print(float(x))     # str → float (only works if string is numeric)
```

```python
x = True  # Boolean data type stored inside variable x
print(float(x))     # bool → float (True → 1.0, False → 0.0)
```

```python
x = False  # Boolean data type stored inside variable x
print(float(x))     # bool → float (True → 1.0, False → 0.0)
```

### bool() Function

```python
x = 5  # Integer data type stored inside variable x
print(bool(x))      # int → bool
```

```python
x = 0  # Integer data type stored inside variable x
print(bool(x))      # int → bool
```

```python
x = 3.14  # Float data type stored inside variable x
print(bool(x))      # float → bool
```

```python
x = 0.0  # Float data type stored inside variable x
print(bool(x))      # float → bool
```

```python
x = "Hello"  # String data type stored inside variable x
print(bool(x))      # str → bool (False only if string is empty)
```

```python
x = ""  # String data type stored inside variable x
print(bool(x))      # str → bool (False only if string is empty)
```

```python
x = [1, 2, 3]  # List data type stored inside variable x
print(bool(x))      # list → bool (False only if list is empty)
```

```python
x = []  # List data type stored inside variable x
print(bool(x))      # list → bool (False only if list is empty)
```

### list() Function

```python
x = "Python"  # String data type stored inside variable x
print(list(x))       # str → list of characters
```
