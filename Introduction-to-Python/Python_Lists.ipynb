# Subsetting Lists

```python
# Python List
family_name_weights = ["Jack", 100, "Zyler", 50, "Vinod", 75, "Ahmed", 60]
```

```python
# Subsetting Index 0 Element
print(family_name_weights[0])
```

```python
# Subsetting Index 4 Element
print(family_name_weights[4])
```

```python
# Subsetting Index -1 Element
print(family_name_weights[-1])
```

```python
# Subsetting Index 7 Element
print(family_name_weights[7])
```

```python
# Slicing List
print(family_name_weights[0:5])
```

```python
# Slicing List
print(family_name_weights[1:4])
```

```python
# Slicing List without start Index
print(family_name_weights[:4])
```

```python
# Slicing List without end Index
print(family_name_weights[4:])
```

# Manipulating Lists

```python
# Vinod has gained some weight
family_name_weights[5] = 80
```

```python
# Jack has changed his name to Zac
family_name_weights[0] = "Zac"
```

```python
# Ahmed has reduced some weight
family_name_weights[-1] = 55
```

```python
# printing List
print(family_name_weights)
```

```python
# Mathematics Marks
maths = [50, 60, 55]

# CS Marks
cs = [91, 28, 32]

# Combined List
maths_cs = maths + cs

# printing list
print(maths_cs)
```

```python
# Deleting Index 2 Element
del maths_cs[2]
```

```python
# Deleting Index -4 Element
del maths_cs[-4]
```

```python
# Deleting Index 1 uptill Index 2 Element
del maths_cs[1:3]
```

```python
# printing List
print(maths_cs)
```

```python
# Create list areas
areas = [11.25, 18.0, 20.0, 10.75, 9.50]
```

```python
# Mirror List
areas_mirror = areas
```

```python
# Changing element in areas
areas[0] = 10
```

```python
# Checking out areas_mirror
print(areas_mirror)
```

```python
# Slicing entire List
areas_copy = areas[:]
```

```python
# Using list() Function
areas_copy2 = list(areas)
```

```python
print(areas_copy)
```
