# Introduction to NumPy

```python
# 2 Lists
height = [1.73, 1.68, 1.71]
weight = [65.4, 59.2, 63.6]
```

```python
# Quick BMI Calculation is impossible
# weight/(height**2) --> throws an Error
```

```python
# Importing NumPy Package
import numpy as np
```

```python
# Converting Lists into ndArrays
np_height = np.array(height)
np_weight = np.array(weight)
```

```python
# Deriving element wise BMI
bmi = np_weight / (np_height**2)
print(bmi)
```

```python
# Heterogeneous List
l = [1,2,3.5,"hi"]

# converting into ndArray
np_l = np.array(l)
print(np_l)
```

```python
# List and ndArray
python_list = [1,2,3]
numpy_array = np.array([1,2,3])

# Lists appends when added
print(python_list + python_list)

# ndArrays add when added
print(numpy_array + numpy_array)
```

```python
# ndArray of Scores
scores = np.array([23, 54, 75, 89, 12, 29])
```

```python
# indexing Element 0
print(scores[0])
```

```python
# Slicing Element 1 to 4
print(scores[1:5])
```

```python
# Comparison Operator
scores>50
```

```python
# Elements greater than 50
print(scores[scores>50])
```

# 2D Arrays

```python
# type of NumPy Array
type(scores)
```

```python
# Creating 2D Arrays with 3 Lists inside a List
np_2d = np.array([[55, 21, 23, 89, 51], [12, 25, 56, 78, 45], [53, 22, 47, 58, 99]])
```

```python
# Applying .shape method
np_2d.shape
```

```python
# Printing Array
print(np_2d)
```

```python
# Indexing Row 0
print(np_2d[0])
```

```python
# Indexing Row 1
print(np_2d[1])
```

```python
# Indexing Row 2
print(np_2d[2])
```

```python
# Indexing Scaler at Row 0 Column 1
print(np_2d[0,1])
```

```python
# Indexing Scaler at Row 2 Column 3
print(np_2d[2,3])
```

```python
# Indexing Scaler at Row 1 Column 2
print(np_2d[1,2])
```

```python
# Slicing Rows 0 to 1
print(np_2d[0:2])
```

```python
# Slicing Rows from Start till End
print(np_2d[:])
```

```python
# Slicing R0 to R1, C1 to C2
print(np_2d[0:2, 1:3])
```

```python
# Slicing R1 to End, Start to C1
print(np_2d[1:, :2])
```

```python
# Indexing and Slicing
print(np_2d[0,1:4])
```

```python
# Indexing and Slicing
print(np_2d[0:2,1])
```

```python
# All Elements greater than 50
print(np_2d[np_2d>50])
```

```python
# All Even Elements
print(np_2d[(np_2d%2)==0])
```

# Basic Statistics

```python
# Sum of all Elements
print(np.sum(np_2d))
```

```python
# Max out of all Elements
print(np.max(np_2d))
```

```python
# Min out of all Elements
print(np.min(np_2d))
```

```python
# Average of all Elements
print(np.mean(np_2d))
```

```python
# St. Dev of all Elements
print(np.std(np_2d))
```

```python
# Variance of all Elements
print(np.var(np_2d))
```

```python
# Max of Index 1 Row
print(np.max(np_2d[1]))
```

```python
# Min of Index 2 Column
print(np.min(np_2d[:,2]))
```

```python
# Squash the Rows
np.max(np_2d, axis=0)
```

```python
# Squash the Columns and don't drop dimension
np.min(np_2d, axis=1, keepdims=True)
```
