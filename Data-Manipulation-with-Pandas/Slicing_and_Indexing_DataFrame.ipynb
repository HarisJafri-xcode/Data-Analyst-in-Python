# Explicit Indexes

```python
# Loading Important Data Science Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Storing file path
file_path = "https://raw.githubusercontent.com/HarisJafri-xcode/Python-Session-DataSets/refs/heads/main/Data_Manipulation/temperatures.csv"
```

```python
# Loading Dataset into a DataFrame
temperatures = pd.read_csv(file_path)
```

```python
# Observing first 5 Rows of the Data Frame
temperatures.head()
```

```python
# Observing bottom 5 Rows of the Data Frame
temperatures.tail()
```

```python
# Extracting Information about Data Frame
temperatures.info()
```

```python
# Converting date column to datetime object
temperatures['date'] = pd.to_datetime(temperatures['date'],format='%d/%m/%Y')
```

```python
# Setting 'city' as Index
temperatures_ind = temperatures.set_index('city')
temperatures_ind.head()
```

```python
# Resetting Index without Deleting the Column
temperatures_ind.reset_index()
```

```python
# Resetting Index after Deleting the Column
temperatures_ind.reset_index(drop=True)
```

```python
# Sorting Dataframe by Index in Descending Order
temperatures_ind.sort_index(ascending=False)
```

# .loc vs .iloc

```python
# Label based Subsetting
temperatures_ind.loc["Xian",["date","country"]]
```

```python
# Index based Subsetting
temperatures_ind.iloc[-165:,0:2]
```
