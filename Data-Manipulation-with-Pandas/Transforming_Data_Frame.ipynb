# Sorting and Subsetting Dataframe

```python
# Sorting Data Frame based on "individuals"
homelessness_ind = homelessness.sort_values('individuals')
homelessness_ind.head()
```

```python
# Sorting Data Frame based on "family_members" based on descending order
homelessness_fam = homelessness.sort_values('family_members',ascending=False)
homelessness_fam.head()
```

```python
# Sorting on Multiple Columns
homelessness_reg_fam = homelessness.sort_values(['region','family_members'],ascending=[True,False])
homelessness_reg_fam.head()
```

```python
# Displaying Top 5 Rows of Column Data Series
homelessness['individuals'].head()
```

```python
# Storing the Data Series inside a Variable
individuals = homelessness['individuals']
individuals.head()
```

```python
# Extracting DataFrame from Data Frame
state_fam = homelessness[['state','family_members']]
state_fam.head()
```

```python
# Extracting DataFrame from Data Frame
ind_state = homelessness[['individuals','state']]
ind_state.head()
```

```python
# Constructing Mask and use it to Subset Data Frame
ind_gt_10k_mask = homelessness['individuals'] > 10000
ind_gt_10k = homelessness[ind_gt_10k_mask]
ind_gt_10k
```

```python
# Constructing Mask and use it to Subset Data Frame
mountain_reg_mask = (homelessness['region'] == 'Mountain')
mountain_reg = homelessness[mountain_reg_mask]
mountain_reg
```

```python
# Constructing Multiple Masks
fam_lt_1k_pac_condition_1 = homelessness['family_members'] < 1000
fam_lt_1k_pac_condition_2 = homelessness['region'] == "Pacific"
```

```python
# Passing Mask to filter rows that comply both conditions
fam_lt_1k_pac = homelessness[fam_lt_1k_pac_condition_1 & fam_lt_1k_pac_condition_2]
fam_lt_1k_pac
```

```python
# Subsetting Rows by Categorical Variables
important_states = ["California", "Arizona", "Nevada", "Utah"]

# Creating Mask
mask_important_states = homelessness['state'].isin(important_states)

# Filter for rows
important_states_homelessness = homelessness[mask_important_states]

# Display those important states
important_states_homelessness
```

# New Columns

```python
# Add total col as sum of individuals and family_members
homelessness['total'] = homelessness['individuals'] + homelessness['family_members']

# display homelessness
homelessness.head()
```

```python
# Add p_homeless col as proportion of total homeless population to the state population
homelessness['p_homeless'] = (homelessness['total']/homelessness['state_pop'])*100

# display homelessness
homelessness.head()
```

Which state has the highest number of homeless individuals per 10,000 people in the state?

```python
# Creating a New Column to find how many homeless people are there for every 10,000 people in each state
homelessness["indiv_per_10k"] = 10000 * (homelessness['individuals'] / homelessness['state_pop'] )

# Display Dataframe
homelessness.head()
```

```python
# Sort the Dataframe with respect to indiv_per_10k Column in ascending order and storing it
homelessness_srt = homelessness.sort_values('indiv_per_10k',ascending=False)
```
