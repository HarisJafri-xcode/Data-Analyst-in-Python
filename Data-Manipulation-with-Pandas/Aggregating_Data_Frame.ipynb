# Summary Statistics


```python
# Loading Important Data Science Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Storing file path
file_path = "https://raw.githubusercontent.com/HarisJafri-xcode/Python-Session-DataSets/refs/heads/main/Data_Manipulation/sales.csv"
```

```python
# Loading Dataset inside Dataframe
sales = pd.read_csv(file_path, index_col=0)
```

```python
# displaying first few rows of the DataFrame
sales.head()
```

```python
# Info about the sales DataFrame
sales.info()
```

```python
# Mean of Weekly sales column
sales['weekly_sales'].mean()
```

```python
# printing Mean of Weekly sales column
print(sales['weekly_sales'].mean())
```

```python
# Median of Weekly sales column
sales['weekly_sales'].median()
```

```python
# Converting Date Column to pandas datetime dtype
sales['date'] = pd.to_datetime(sales['date'],format='%Y-%m-%d')
```

```python
# Print the maximum of the date column
print(sales['date'].max())
```

```python
# Print the minimum of the date column
print(sales['date'].min())
```

```python
# A custom IQR function
def iqr(column):
    return column.quantile(0.75) - column.quantile(0.25)

# Display IQR of the temperature_c column
sales['temperature_c'].agg(iqr)
```

```python
# Applying iqr Function on 3 Columns
sales[["temperature_c", 'fuel_price_usd_per_l', 'unemployment']].agg(iqr)
```

```python
# Applying iqr, median and mean Function on 3 Columns
sales[["temperature_c", "fuel_price_usd_per_l", "unemployment"]].agg([iqr, "median","mean"])
```

```python
# Subsetting Dataframe where Store == 1 & Department == 1
mask = (sales['store'] == 1) & (sales['department'] == 1)
sales_1_1 = sales[mask].copy()
```

```python
# Sorting Via Date
sales_1_1 = sales_1_1.sort_values('date')
```

```python
# Adding a New Column
sales_1_1['cum_weekly_sales'] = sales_1_1['weekly_sales'].cumsum()
```

```python
# Displaying Dataframe with 3 Columns
sales_1_1[["date", "weekly_sales", "cum_weekly_sales"]]
```

# Grouped Summary Statistics

## What percent of sales occurred at each store type?

```python
# Calc total weekly sales
sales_all = sales["weekly_sales"].sum()

# Subset for type A stores, calc total weekly sales
sales_A = sales[sales["type"] == "A"]["weekly_sales"].sum()

# Subset for type B stores, calc total weekly sales
sales_B = sales[sales["type"] == "B"]["weekly_sales"].sum()

# Subset for type C stores, calc total weekly sales
sales_C = sales[sales["type"] == "C"]["weekly_sales"].sum()

# Get proportion for each type
sales_propn_by_type = [sales_A, sales_B, sales_C] / sales_all
print(sales_propn_by_type)
```

```python
# Group by type; calc total weekly sales
sales_by_type = sales.groupby("type")["weekly_sales"].sum()

# Get proportion for each type
sales_propn_by_type = sales_by_type / sales_by_type.sum()
print(sales_propn_by_type)
```

## Multiple Grouped Summaries

```python
# Grouping by Type and finding multiple Statistics
sales_stats = sales.groupby("type")["weekly_sales"].agg(["min", "max", "mean", "median"])
sales_stats
```

```python
# Grouping by Type and finding multiple Statistics
unemp_fuel_stats = sales.groupby('type')[['unemployment', 'fuel_price_usd_per_l']].agg(['mean', 'median'])
unemp_fuel_stats
```
