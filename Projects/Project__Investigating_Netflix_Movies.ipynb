# Project Solution

```python
# @title
# Extracting Rows where type is Movie
netflix_df_movie = netflix_df[netflix_df["type"] == "Movie"]

# Extracting Rows where movie was released between 1990 and 1999
netflix_df_movie_90s = netflix_df_movie[(netflix_df_movie["release_year"] >= 1990) & (netflix_df_movie["release_year"] <= 1999)]

# Save the value counts to a variable
counts = netflix_df_movie_90s["duration"].value_counts()

# Extract the first index (the duration) and the first value (the frequency)
duration = counts.index[0]
highest_frequency = counts.iloc[0]

# Print the dynamically generated statement
print(f"{duration} minute movies occured the highest {highest_frequency} times !")

# short movies of 90s
netflix_df_movie_90s_short = netflix_df_movie_90s[netflix_df_movie_90s["duration"] < 90]

# short action movies of 90s
netflix_df_movie_90s_short_action = netflix_df_movie_90s_short[netflix_df_movie_90s_short["genre"] == "Action"]

# How many Rows ?
short_movie_count = netflix_df_movie_90s_short_action.shape[0]

# Print the dynamically generated statement
print("Total Action Movies with runtime less than 90 minutes: ",short_movie_count)
```
