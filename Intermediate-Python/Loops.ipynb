# while Loop

```python
# Declaring Error
error = 100
threshold = 50

# Show Warning until Error is below Threshold
while error > threshold:
    print("Warning")
    error = error - 20
print("No Need to Worry")
```

# for Loop

```python
# String
name = "Haris"

# Using Index
print(name[0])
print(name[1])
print(name[2])
print(name[3])
print(name[4])
```

```python
# for Loop
for char in name:
    print(char)
```

```python
# List
age = [12, 13, 11]

# for Loop
for num in age:
    print(num)
```

```python
# Family Age
ages = [12, 13, 11]

# for Loop
for index, age in enumerate(ages):
    print("Person", index+1, "shall be", age+2, "years old after 2 Years")
```

```python
# house list of lists
house = [["hallway", 11.25],
         ["kitchen", 18.0],
         ["living room", 20.0],
         ["bedroom", 10.75],
         ["bathroom", 9.50]]

# Build a for loop from scratch
for x in house:
    print("the", x[0], "is", x[1], "sqm")
```

# Loops with other Data Structures

```python
# Definition of dictionary
europe = {'spain':'madrid', 'france':'paris', 'germany':'berlin',
          'norway':'oslo', 'italy':'rome', 'poland':'warsaw', 'austria':'vienna' }
```

```python
# Using Loop
for key, value in europe.items():
    print("Capital of", key.capitalize() , "is", value.capitalize())
```

```python
import numpy as np
```

```python
# 2D NumPy Array
numbers = np.array([[12, 25 ,13], [7, 18, 9]])
```

```python
# Iteration
for n in numbers:
    print(n)
```

```python
# Iteration
for n in np.nditer(numbers):
    print(n)
```

```python
import pandas as pd
```

```python
cars_df = pd.read_csv('https://raw.githubusercontent.com/HarisJafri-xcode/Python-Session-DataSets/refs/heads/main/Intermediate_Python/cars.csv')
```

```python
# applying for Loop
for c in cars_df:
    print(c)
```

```python
# Current Pandas Dataframe
cars_df
```

```python
# Adding a New Column
cars_df["check_column"] = True
```

```python
# Revised Pandas DF
cars_df
```

```python
# deleting a Column
del cars_df["check_column"]
```

```python
# Checking again after Column deletion
cars_df
```

```python
# Creating a Column to identify high cars_per_cap countries
cars_df["high_cars_per_cap"] = cars_df["cars_per_cap"] > 200
```

```python
# df after column addition
cars_df
```

```python
# Using .iloc to extract first three rows of country column
cars_df["country"].iloc[0:3]
```

```python
# Using .loc to extract first three rows of country column
cars_df.loc[0:2, "country"]
```

```python
# Using .iloc to extract first three rows of country and drives_right column
cars_df[["country","drives_right"]].iloc[0:3]
```

```python
# Using .loc to extract first three rows of country and drives_right column
cars_df.loc[0:2, ["country","drives_right"]]
```

```python
# iterating over each row individually
for index, row in cars_df.iterrows():
    print(index)
    print(row)
```

```python
# Creating a new Column
cars_df["country_code"] = ""
```

```python
# Filling Country Code
for index, row in cars_df.iterrows():
    cars_df.loc[index, "country_code"] = cars_df.loc[index, "country"][:2]
```
