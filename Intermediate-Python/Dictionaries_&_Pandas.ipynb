# Pandas

```python
# importing pandas
import pandas as pd
```

```python
# Predefined lists
names = ['United States', 'Australia', 'Japan', 'India', 'Russia', 'Morocco', 'Egypt']
drives_right =  [True, False, False, False, True, True, True]
cars_per_cap = [809, 731, 588, 18, 200, 70, 45]
```

```python
# Create dictionary cars_dict with three key:value pairs
# Keys are "String" and Column Heading
# Values are "Lists" and Column Content
cars_dict = {'country':names,'drives_right':drives_right,'cars_per_cap':cars_per_cap}
```

```python
print(cars_dict)
```

```python
cars_df = pd.DataFrame(cars_dict)
```

```python
cars_df
```

```python
print(cars_df)
```
