# Logical Operators

```python
# Define variables
my_kitchen = 18.0
your_kitchen = 14.0
```

```python
# my_kitchen bigger than 10 and smaller than 18?
print(my_kitchen>10 and my_kitchen<18)
```

```python
# my_kitchen smaller than 14 or bigger than 17?
print(my_kitchen<14 or my_kitchen>17)
```

```python
# Double my_kitchen smaller than triple your_kitchen?
print(not(my_kitchen < your_kitchen))
```

```python
# Complex Situation
x = 8
y = 9
not(not(x < 3) and not(y > 14 or y > 10))
```

```python
# Create arrays
import numpy as np
my_marks = np.array([18.0, 20.0, 10.75, 9.50])
```

```python
# Boolean Array - Marks > 15
print(my_marks > 15)
```

```python
# Using Boolean Array to filter
my_marks[my_marks > 15]
```

```python
# Boolean Array with marks > 15 and marks < 20
print(my_marks[(my_marks > 15) & (my_marks < 20)])
```

# if, elif and else

```python
# Declaring a Variable
points = 10

# Checking if point is greater than 25
if (points > 25):
    print("You have earned more than 25 points")
```

```python
# Checking if point is greater than 5
if (points > 5):
    print("You have earned more than 5 points")
```

```python
# Declaring Variable
eggs = 8

# are eggs greater than 10 ?
if (eggs > 10):
    print("There are more than 10 eggs")

# are eggs greater than 5 ?
if (eggs > 5):
    print("There are more than 5 eggs")
```

```python
# Declaring Variable
num = 12

# Nested ifs
if (num>6):
    print("num is greater than 6")
    if (num>20):
        print("num is greater than 20")
        if (num<25):
            print("num is less than 25")
```

```python
# Declaring Variable
num = 12

# if else Statement
if (num<6):
    print("num is greater than 6")
else:
    print("num is less than 6")
```

```python
# if else Statement

if (1+1 == 2):
    print("True Statement")
else:
    print("False Statement")
```

```python
# if else Statement

if (1+1 == 2):
    if (2+2 == 5):
        print("Yes")
    else:
        print("No")
else:
    print("We don't know")
```

```python
# if else elif
num = 5

if (num == 1):
    print("num is 1")
elif (num == 2):
    print("num is 2")
elif (num == 3):
    print("num is 3")
elif (num == 4):
    print("num is 4")
else:
    print("num is greater than 4")
```

```python
# Declare two variables
temperature = 25
is_sunny = True

# Use if, elif, else with logical operators
if (temperature > 30) and is_sunny:
    print("It's a hot and sunny day, perfect for the beach!")
elif (temperature <= 30) and is_sunny and not(temperature < 15):
    print("It's a pleasant and sunny day.")
elif (temperature < 15) or not is_sunny:
    print("It's a bit chilly or not sunny. Maybe wear a jacket.")
else:
    print("Weather conditions are a bit unusual today!")
```

# Filtering Pandas DataFrame

```python
import numpy as np
import pandas as pd
```

```python
# Loading DataFrame from CSV
cars_df = pd.read_csv('https://raw.githubusercontent.com/HarisJafri-xcode/Python-Session-DataSets/refs/heads/main/Intermediate_Python/cars.csv')
```

```python
# Information about DataFrame
cars_df.info()
```

```python
# View df
cars_df
```

```python
# cars_per_cap Data Series
cars_df["cars_per_cap"]
```

```python
# cars_per_cap greater than 100
cars_df["cars_per_cap"] > 100
```

```python
# Storing Boolean DS
mask = cars_df["cars_per_cap"] > 100

# Rows where cars_per_cap is greater than 100
cars_df[mask]
```

```python
# cpc > 100
cpc_mask = cars_df["cars_per_cap"] > 100

# drives right
drive_mask = cars_df["drives_right"] == True

# Subset
cars_df[cpc_mask | drive_mask]
```

```python
# Subset
cars_df[cpc_mask & drive_mask]
```
