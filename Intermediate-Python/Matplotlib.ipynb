# Basic Plots with Matplotlib

```python
# Importing subplot
import matplotlib.pyplot as plt
```

```python
# Two Lists
year = [1950, 1970, 1990, 2010]
pop = [2.519, 3.692, 5.263, 6.972]
```

```python
# Constructing and Displaying Line Chart
plt.plot(year, pop)
plt.show()
```

```python
# Constructing and Displaying Scatter Plot
plt.scatter(year, pop)
plt.show()
```

# Histogram

```python
# Taking help on plt.hist()
help(plt.hist)
```

```python
# Creating an array
values = [0, 0.6, 1.4, 1.2, 2.7, 2.8, 1.4, 2.4, 3.7, 1.2, 3.1, 1.8, 5.1]
```

```python
# Constructing and Displaying Histogram
plt.hist(values)
plt.show()
```

```python
# Constructing and Displaying Histogram with 5 Bins
plt.hist(values, bins=5)
plt.show()
```

# Customization

```python
# Two Lists
year = [1950, 1970, 1990, 2010]
pop = [2.519, 3.692, 5.263, 6.972]

# Visualizing
plt.plot(year, pop)
plt.show()
```

```python
# Two Lists
year = [1950, 1970, 1990, 2010]
pop = [2.519, 3.692, 5.263, 6.972]

# Visualizing
plt.plot(year, pop)
plt.xlabel('Year')
plt.ylabel('Population')
plt.title("World Population Projections")
plt.show()
```

```python
# Two Lists
year = [1950, 1970, 1990, 2010]
pop = [2.519, 3.692, 5.263, 6.972]

# Adding New Data to List
year = [1930, 1940] + year + [2020, 2030]
pop = [2.439, 2.507] + pop + [10.732, 13.007]

# Visualizing
plt.plot(year, pop)
plt.xlabel('Year')
plt.ylabel('Population')
plt.title("World Population Projections")
plt.show()
```

```python
# Two Lists
year = [1950, 1970, 1990, 2010]
pop = [2.519, 3.692, 5.263, 6.972]

# Adding New Data to List
year = [1930, 1940] + year + [2020, 2030]
pop = [2.439, 2.507] + pop + [10.732, 13.007]

# Visualizing
plt.plot(year, pop)
plt.xlabel('Year')
plt.ylabel('Population')
plt.title("World Population Projections")
plt.yticks([0, 2, 4, 6, 8, 10, 12, 14], ['0B', '2B', '4B', '6B', '8B', '10B', '12B', '14B'])
plt.show()
```
